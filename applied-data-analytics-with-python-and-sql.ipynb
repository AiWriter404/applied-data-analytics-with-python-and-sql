# **3- Operators in Python**

```python
# Arethmatic Operators
a=10
b=3
# Addition
print("addition:",a+b)

# Substraction
print("Substraction:",a-b)

# Multiplication

print("Multiplication:",a*b)

# Division

print("Division",a/b)

# Floor Division (Remove decimal part)

print("Floor Division:",a//b)

# Modulus (Remainder)

print("Modulus:",a%b)

# Exponentiation (Power)

print("Exponentiation:",a**b)

```

```python
# Comparison Operators

x=5
y=10

# Equal to

print("Equal to :", x==y)

# Not Equal 

print("Not Equal:", x!=y)

# Greater Than

print("Greater Than:",x > y)

# Greater than or Equal to 

print("Greater Than Or Equal to:",x >= y)

# Less Than or Equal to 

print("Less Than And Equal to :", x <= y)
```

```python
# Assignment Operators

# Add  and assign

x+=5

print("x after +=5:", x)

# Substract and Assign

x -=3
print("x after -=3:",x)

# Multiply and Assign

x *=2
print("x after *=2:",x)

# Divide and Assign

x /=4
print("x after /=4:", x)
```

```python
# Logical Operators
a = True
b = False

# Logical AND

print("a AND b:",a and b)

# Logical OR

print("a OR b:",a or b)

# Logical NOT

print("NOT a:", not a)
```

```python
# Bitwise operators
x = 5  # 0b0101 (binary)
y = 3  # 0b0011 (binary)

# AND
print("x & y (AND):", x & y)  # Result: 1 (0b0001)

# OR
print("x | y (OR):", x | y)  # Result: 7 (0b0111)

# XOR
print("x ^ y (XOR):", x ^ y)  # Result: 6 (0b0110)

# NOT
print("~x (NOT):", ~x)  # Result: -6 (two's complement)

# Left Shift
print("x << 1 (Left Shift):", x << 1)  # Result: 10 (0b1010)

# Right Shift
print("x >> 1 (Right Shift):", x >> 1)  # Result: 2 (0b0010)
```

```python
# Membership operators
my_list = [1, 2, 3, 4, 5]

# In operator
print("Is 3 in the list?", 3 in my_list)

# Not in operator
print("Is 10 not in the list?", 10 not in my_list)
```

```python
# Identity operators
x = [1, 2, 3]
y = [1, 2, 3]
z = x

# Is operator
print("x is z:", x is z)

# Is not operator
print("x is not y:", x is not y)
```

- Arithmetic operators for basic math operations,
- Comparison operators to compare values,
- Assignment operators for assigning values,
- Logical operators to combine conditions,
- Bitwise operators for bit-level operations,
- Membership operators to check if a value is in a sequence,
- Identity operators to compare memory addresses.
