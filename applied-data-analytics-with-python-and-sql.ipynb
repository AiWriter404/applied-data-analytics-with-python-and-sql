# **1- Identifiers in Pyhton**

```python
# Defining Variables (Identifiers) (Variables)
age=25
name="alice"
#Defining a function (Identifiers)
def greet():
    print("Hello World")
# Calling a function
greet()
```

1. Must Begin with letters (a to z) or (A to Z) or underscore(_)
2. cannot start with a number
3. can only contains alphanumeric characters(letter, numbers and underscores)
4. Case sensitive : Name and name are two different Identifiers .
5. cannot use python keywords as identifiers (as like for if else etc)
